```python
import sklearn

print(sklearn.__version__)
```

```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split

import pandas as pd
```

```python
# 붓꽃 데이터 세트 로딩
iris = load_iris()

iris
```

```python
# 데이터 프레임 생성
data = pd.DataFrame(data=iris['data'], columns=iris['feature_names'])
target = pd.DataFrame(data=iris['target'], columns=['target'])

data
```

```python
target
```

```python
df = pd.concat([data, target], axis=1)

df.to_csv('iris.csv', index=False)

df
```

```python
df = pd.read_csv('iris.csv')

df
```

```python
# 학습 데이터, 테스트 데이터 분류
X_train, X_test, y_train, y_test = train_test_split(data, target, test_size=0.2, random_state=11)
```

```python
X_train
```

```python
X_test
```

```python
y_train
```

```python
y_test
```

```python
# 모델 생성
dt_clf = DecisionTreeClassifier(random_state=11)
```

```python
# 학습 수행
dt_clf.fit(X_train, y_train)
```

```python
# 예측
pred = dt_clf.predict(X_test)

pred
```

```python
y_test
```

```python
pred
```

```python
from sklearn.metrics import accuracy_score
print('예측 정확도: {0:.4f}'.format(accuracy_score(y_test, pred)))
```

```python
# 예측하기
new_data = [[5.1, 3.0, 3, 1.2]]
new_pred = dt_clf.predict(new_data)

new_pred
```

```python
# 학습 모델 덤프
import joblib
joblib.dump(dt_clf, 'model.joblib')
```

```python
# 학습 모델 로드(불러오기)
load_model = joblib.load('model.joblib')
```

```python
# 예측하기
load_model.predict(new_data)
```
